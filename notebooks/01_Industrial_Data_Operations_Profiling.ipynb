##2.Kode mengekstrak file CSV ke folder data/raw tanpa mengubah isi datanya

```python
import zipfile
from pathlib import Path

# Folder penyimpanan project
PROJECT_DIR = Path("/content/industrial-data-operations")
RAW_DIR = PROJECT_DIR / "data" / "raw"
REPORT_DIR = PROJECT_DIR / "reports"

RAW_DIR.mkdir(parents=True, exist_ok=True)
REPORT_DIR.mkdir(parents=True, exist_ok=True)

# Cari file ZIP yang diupload
zip_files = [
    name for name in uploaded.keys()
    if name.lower().endswith(".zip")
]

if not zip_files:
    raise ValueError("File ZIP tidak ditemukan. Upload ZIP dataset terlebih dahulu.")

# Ekstrak file CSV ke folder raw
with zipfile.ZipFile(zip_files[0], "r") as z:
    csv_files = [
        name for name in z.namelist()
        if name.lower().endswith(".csv")
        and not name.startswith("__MACOSX/")
    ]

    for name in csv_files:
        destination = RAW_DIR / Path(name).name

        with z.open(name) as source:
            destination.write_bytes(source.read())

print("Jumlah file CSV:", len(csv_files))
print("\nFile yang ditemukan:")

for path in sorted(RAW_DIR.glob("*.csv")):
    print(path.name, "-", round(path.stat().st_size / 1024, 2), "KB")
```

##3.Import library dan baca CSV

Membaca semua CSV terlebih dahulu karena nama file di beberapa versi dataset mungkin berbeda.

```python
import pandas as pd
import numpy as np

print("Pandas:", pd.__version__)
print("NumPy:", np.__version__)

# Membaca semua file CSV
data = {}

for path in sorted(RAW_DIR.glob("*.csv")):
    name = path.stem
    data[name] = pd.read_csv(path)

    print(f"{name}: berhasil dibaca")

print("\nTotal dataset:", len(data))
```

```python
quality_results = []

for name, df in data.items():
    quality_results.append({
        "dataset": name,
        "rows": len(df),
        "columns": len(df.columns),
        "missing_values": int(df.isna().sum().sum()),
        "duplicate_rows": int(df.duplicated().sum()),
        "memory_mb": round(
            df.memory_usage(deep=True).sum() / (1024 ** 2), 2
        )
    })

quality_report = pd.DataFrame(quality_results)

display(quality_report)
```

##3.1 periksa kelengkapan telemetry
Telemetry memiliki 876.100 baris. Kita perlu memeriksa apakah setiap mesin memiliki jumlah observasi yang sama dan apakah ada interval waktu yang hilang

```python
# Gunakan salinan data agar raw data tetap aman
telemetry_check = data["PdM_telemetry"].copy()

telemetry_check["datetime"] = pd.to_datetime(
    telemetry_check["datetime"],
    errors="coerce"
)

# Jumlah observasi setiap mesin
observations = (
    telemetry_check
    .groupby("machineID")
    .agg(
        total_records=("datetime", "size"),
        first_reading=("datetime", "min"),
        last_reading=("datetime", "max"),
        unique_timestamps=("datetime", "nunique")
    )
    .reset_index()
)

display(observations.head(10))

print("Jumlah mesin:", len(observations))
print("Record minimum:", observations["total_records"].min())
print("Record maksimum:", observations["total_records"].max())
print(
    "Mesin dengan jumlah record berbeda:",
    observations["total_records"].nunique()
)
```

##Interpretasi hasil Telemetry

- 100 mesin tercatat dalam master data.
- Setiap mesin memiliki 8.761 record.
- Record paling sedikit dan paling banyak sama-sama 8.761.
- Setiap mesin memiliki 8.761 timestamp unik.
- Semua mesin memiliki waktu pengamatan awal 2015-01-01 06:00:00 dan akhir 2016-01-01 06:00:00.

Totalnya adalah \(100 x 8.761 = 876.100\) baris, sesuai dengan jumlah data telemetry sebelumnya.
Ini indikasi kuat bahwa setiap mesin memiliki jadwal pencatatan yang seragam. Namun, kita belum membuktikan bahwa tidak ada jam yang terlewat di tengah periode. Kita perlu memeriksa interval waktunya.

## 3.2 cek apakah ada timestamp yang terlewat

mengecek apakah jarak antarobservasi setiap mesin selalu satu jam.

```python
# Urutkan data berdasarkan mesin dan waktu
telemetry_check = telemetry_check.sort_values(
    ["machineID", "datetime"]
).copy()

# Hitung selisih waktu antarrecord untuk setiap mesin
telemetry_check["time_gap"] = (
    telemetry_check.groupby("machineID")["datetime"].diff()
)

# Ringkasan interval waktu
print("Distribusi interval waktu:")
display(
    telemetry_check["time_gap"]
    .value_counts()
    .rename_axis("time_interval")
    .reset_index(name="count")
    .head(10)
)

# Periksa interval yang bukan satu jam
unexpected_gaps = telemetry_check[
    telemetry_check["time_gap"].notna()
    & (telemetry_check["time_gap"] != pd.Timedelta(hours=1))
]

print("\nJumlah interval yang bukan satu jam:", len(unexpected_gaps))

display(
    unexpected_gaps[
        ["machineID", "datetime", "time_gap"]
    ].head(10)
)
```

##Pemeriksaan interval waktu
hasilnya tidak ditemukan interval yang bukan satu jam

-Total interval satu jam: 876.000

-Interval yang tidak sesuai: 0

-Setiap mesin memiliki 8.761 observasi.

-Ada 100 mesin dengan rentang waktu yang konsisten

sehingga untuk seluruh interval antarobservasi yang diperiksa, telemetry tercatat setiap satu jam tanpa jeda yang tidak biasa

##3.3 Periksa nilai sensor
Sekarang kita akan mengecek apakah ada nilai sensor yang terlalu ekstrem atau tidak masuk akal. Ini penting sebelum masuk ke EDA dan feature engineering

```python
# Kolom sensor yang akan diperiksa
sensor_cols = ["volt", "rotate", "pressure", "vibration"]

# Statistik dasar
sensor_summary = telemetry_check[sensor_cols].describe().T

# Tambahkan jumlah nilai unik dan nilai nol
sensor_summary["unique_values"] = (
    telemetry_check[sensor_cols].nunique()
)

sensor_summary["zero_count"] = (
    telemetry_check[sensor_cols].eq(0).sum()
)

display(sensor_summary)
```

periksa nilai ekstrem dengan :
metode IQR

```python
outlier_results = []

for col in sensor_cols:
    q1 = telemetry_check[col].quantile(0.25)
    q3 = telemetry_check[col].quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outlier_mask = (
        (telemetry_check[col] < lower_bound)
        | (telemetry_check[col] > upper_bound)
    )

    outlier_results.append({
        "sensor": col,
        "lower_bound": lower_bound,
        "upper_bound": upper_bound,
        "outlier_count": int(outlier_mask.sum()),
        "outlier_percentage": round(
            outlier_mask.mean() * 100, 2
        )
    })

outlier_report = pd.DataFrame(outlier_results)
display(outlier_report)
```

Penting: nilai yang terdeteksi sebagai outlier belum tentu salah. Pada data mesin, nilai ekstrem bisa menunjukkan kondisi operasional yang penting atau indikasi kerusakan. Kita hanya menandainya terlebih dahulu, bukan langsung menghapusnya

Dari empat sensor tersebut, pressure memiliki proporsi outlier paling tinggi, yaitu 1,78%. Ini layak diperiksa lebih lanjut, tetapi kita belum bisa menyimpulkan bahwa pressure adalah sumber masalah operasional terbesar

##3.4 periksa apakah outlier perlu ditindaklanjuti

Kita jangan menghapus outlier.  Data Operations, tunjukkan bahwa kamu bisa mengidentifikasi anomali, memeriksa konteksnya, dan mendokumentasikan keputusan

```python
# Periksa 5 nilai tertinggi dan 5 nilai terendah setiap sensor

for col in sensor_cols:
    print(f"\n{'=' * 45}")
    print(f"SENSOR: {col}")
    print("=" * 45)

    print("\n5 nilai terendah:")
    display(
        telemetry_check
        .nsmallest(5, col)[["machineID", "datetime", col]]
    )

    print("\n5 nilai tertinggi:")
    display(
        telemetry_check
        .nlargest(5, col)[["machineID", "datetime", col]]
    )
```

nilai ekstrem terkonsentrasi pada mesin atau waktu tertentu
